import yfinance as yf
import pandas as pd

data = yf.download("AAPL", start="2020-01-01", end="2025-01-01")

print(data.head())
print(data.columns)

```python
# 1. Calculate daily returns
data["Daily_Return"] = data["Close"].pct_change()

# 2. Calculate 20-day and 50-day moving averages
data["MA20"] = data["Close"].rolling(window=20).mean()
data["MA50"] = data["Close"].rolling(window=50).mean()

# 3. Calculate 20-day volatility
data["Volatility20"] = (
    data["Daily_Return"].rolling(window=20).std()
)

# 4. Calculate 5-day momentum
data["Momentum5"] = data["Close"].pct_change(periods=5)

# 5. Display the results
print(data[[
    "Close",
    "Daily_Return",
    "MA20",
    "MA50",
    "Volatility20",
    "Momentum5"
]].tail(10))
```

```python
print(data.columns.tolist())
```

```python
print("Data type:", type(data))
print("Original columns:", data.columns.tolist())

data["TEST_COLUMN"] = 123

print("Columns after adding test:")
print(data.columns.tolist())
```

```python
data["Daily_Return"] = data["Close"].pct_change()

data["MA20"] = data["Close"].rolling(window=20).mean()

data["MA50"] = data["Close"].rolling(window=50).mean()

data["Volatility20"] = data["Daily_Return"].rolling(window=20).std()

data["Momentum5"] = data["Close"].pct_change(periods=5)

print("ALL COLUMNS:")
print(data.columns.tolist())

print("\nLATEST FEATURE VALUES:")
print(data[[
    "Close",
    "Daily_Return",
    "MA20",
    "MA50",
    "Volatility20",
    "Momentum5"
]].tail(10))
```

```python
data.to_csv("apple_historical_data.csv")

print("Updated dataset saved successfully!")
```

```python
# Predict whether the stock closes higher on the next trading day

data["Next_Close"] = data["Close"].shift(-1)

# Remove the last row because its next-day closing price is unknown
data = data.dropna(subset=["Next_Close"]).copy()

data["Target"] = (
    data["Next_Close"] > data["Close"]
).astype(int)

# Check the result
print(data[["Close", "Next_Close", "Target"]].tail(10))
print("Dataset shape:", data.shape)
```
